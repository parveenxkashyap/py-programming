# #38 Fibonacci Sequence

```python
def fib(n):
    a = 0
    b = 1

    if n == 1:
        print(a)
    elif n <=0:
        print("number should be greater than 0")
    else:

        print(a)
        print(b)

        for i in range(2, n):
            c = a + b
            a = b
            b = c
            print(c)

fib(6)
```

```python
#Change or modify the functionalities of existing functions using decorators

def div(a, b):
    return a/b

def smart_div(func):
    def inner(a,b):
        if a<b:
            a,b=b,a
            return func(a,b)
    return inner



div1 = smart_div(div)
div1(2,4)
```

```python
'''to call the init of both the classes 
by default it is calling the b init if present or if not then init of a class but when both init are there it is only calling the init of b
in order to this to access all the features of superclass including the init there is the method called super()'''
class a: 

    def __init__(self):
        print("In a init")

    def feature1(self):
        print("feature 1 working")
    
    def feature2(self):
        print("feature 2 working")

class b(a): 

    def __init__(self):
        print("In b init")
        super().__init__() #to access all the methods or object of superclass 

    def feature3(self):
        print("feature 3 working")
    
    def feature4(self):
        print("feature 4 working")


a1 = b()
```

# #62 Generators

```python
def topten():

    n = 1

    while n <= 10:
        sq = n*n
        yield sq
        n +=1

values = topten()

# print(values.__next__())
# print(values.__next__())
# print(values.__next__())

for i in values:
    print(i)
```

```python
'''
Thread: lightweigth process, by default every execution has one thread'''

class hello:
    def run(self):
        for i in range(5):
            print("hello")
class hi:
    def run(self):
        for i in range(5):
            print("hi")

t1 = hello()
t2 = hi()

t1.run()
t2.run()
```

```python
from threading import *
from time import sleep
class hello:
    def run(self):
        for i in range(5):
            print("hello")
            sleep(1)
class hi:
    def run(self):
        for i in range(5):
            print("hi")
            sleep(1)

t1 = hello()
t2 = hi()

t1.run()
sleep(0.2)
t2.run()

t1.join()
t2.join()

print("bye")
```

# #69 Binary Search

```python
pos = -1
def search(lst, n):
    global pos
    l = 0
    u = len(lst) - 1

    while l <= u:
        mid = (l + u) // 2
        if lst[mid] == n:
            pos = mid
            return True
        elif lst[mid] < n:
            l = mid + 1
        else:
            u = mid - 1
    return False

# list must be sorted
lst = [1, 2, 3, 4, 5, 8]  
n = 2

if search(lst, n):
    print("found at : ", pos + 1)
else: 
    print("not found")
```

# #70 Bubble Sort 

```python
def sort(nums):
    for i in range(len(nums)-1, 0, -1):
        for j in range(i):
            if nums[j]>nums[j+1]:
                temp = nums[j]
                nums[j] = nums[j+1]
                nums[j+1] = temp
nums = [3,35,2,5,6,3,7,1]

sort(nums)
print(nums)
```
